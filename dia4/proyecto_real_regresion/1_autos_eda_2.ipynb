## 2 CONECTAMOS A BASE DE DATOS Y CARGAMOS DATASET

```python
from sqlalchemy import create_engine
```

```python
from google.colab import userdata
conn_string = userdata.get('CON_STR')
```

```python
# Crear el motor de conexión
try:
    engine = create_engine(conn_string)
    print("Conexión a la base de datos establecida exitosamente.")

    # Cargar la tabla tbl_autos en un DataFrame
    query = "SELECT * FROM tbl_autos"
    df_autos = pd.read_sql(query, engine)

except Exception as e:
    print(f"Error al conectar o cargar datos: {e}")
    engine = None # Asegúrate de que engine sea None si hay un error para evitar problemas posteriores
```

### ANALIZAMOS LAS VARIABLES CATEGORICAS

```python
# GASOLINA
df_clean['fuel'].value_counts()
```

```python
def map_fuel_type(fuel_type):
    if fuel_type == 'Gasolina':
        return 'gasolina'
    else:
        return 'otro'

df_clean['fuel'] = df_clean['fuel'].apply(map_fuel_type)
df_clean['fuel'].value_counts()
```

```python
df_clean['location'].value_counts()
```

```python
def map_location_type(location):
    if location == 'Lima, Lima':
        return 'lima'
    else:
        return 'provincia'

df_clean['location'] = df_clean['location'].apply(map_location_type)
df_clean['location'].value_counts()
```

```python
# marcas
df_clean['brand'].value_counts()
```

```python
df_clean.drop(columns=['brand','location'], inplace=True)
df_clean.head()
```

```python
# year
df_clean['year'].dtype
```

```python
#subcategory
df_clean['subcategory'].value_counts()
```

```python
def map_subcategory_type(subcategory):
    if subcategory == 'Camioneta':
        return 'camioneta'
    elif subcategory == 'Sedán':
        return 'sedan'
    else:
        return 'otros'

df_clean['subcategory'] = df_clean['subcategory'].apply(map_subcategory_type)
df_clean['subcategory'].value_counts()
```

```python
# transmission
df_clean['transmission'].value_counts()
```

```python
def map_transmission_type(transmission):
    if transmission == 'Automática - Secuencial':
        return 'secuencial'
    elif transmission == 'Automática':
        return 'automatica'
    else:
        return 'mecanica'

df_clean['transmission'] = df_clean['transmission'].apply(map_transmission_type)
df_clean['transmission'].value_counts()
```
