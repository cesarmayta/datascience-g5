# IMPORTAMOS LIBRERIA MATPLOTLIB

```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
x = np.linspace(0,5,11)
x
```

```python
y = x**2
y
```

# CREAMOS UN GRAFICO PARA VER LA RELACIÓN ENTRE X Y Y

```python
plt.plot(x,y,color='red',linestyle='--',marker='o',label='y=x^2')
plt.title('Relación entre X y Y')
plt.xlabel('Variable X')
plt.ylabel('Variable Y')
plt.legend()
plt.show()
```

# SUBPLOTS

```python
x = np.linspace(0,5,11)
y = x**2
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,color='red')

plt.subplot(1,2,2)
plt.plot(y,x,color='blue')

plt.show()
```

```python
fig,axes = plt.subplots(2,4)
axes[0,0].plot(x,np.cos(x))
axes[0,1].plot(x,np.sin(x),'yellow')
axes[0,2].plot(x,np.tan(x),'r')
axes[0,3].plot(x,np.cos(x)**2,'g')
axes[1,0].plot(x,np.cos(y),'b')
axes[1,1].plot(x,np.sin(y),'purple')
axes[1,2].plot(x,np.tan(y),'orange')
axes[1,3].plot(x,np.cos(y)**2)

fig.tight_layout()
plt.show()
```

# EJERICIO CON PANDAS

```python
import pandas as pd
df_books = pd.read_csv('https://github.com/cesarmayta/datasets/raw/refs/heads/main/libros.csv')
df_books.head(2)
```

```python
precio_promedio = df_books.groupby('Year')['Price'].mean().reset_index()
precio_promedio
```

```python
x = precio_promedio['Year']
y = precio_promedio['Price']
```

```python
plt.plot(x,y,color='green',label='Evolución del Precio por año')
plt.title('Evolución del Precio promedio por año')
plt.xlabel('Año')
plt.ylabel('Precio promedio')
plt.legend()
plt.show()
```
