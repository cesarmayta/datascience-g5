# IMPORTAMOS LIBRERIAS

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

# CARGAMOS UN DATASET

```python
sns.get_dataset_names()
```

# CARGAMOS UN DATASET EN UN DATAFRAME DE PANDAS

```python
tips = sns.load_dataset('tips')
tips
```

```python
sns.displot(data=tips,x='total_bill')
plt.show()
```

```python
sns.displot(data=tips,x='total_bill',y='tip')
plt.show()
```

# AGRUPACIONES

```python
sns.displot(data=tips,x='total_bill',hue='sex')
plt.show()
```

```python
sns.displot(data=tips,x='total_bill',hue='sex',kind='kde',legend=True,palette='dark')
plt.show()
```

```python
sns.histplot(data=tips,x='tip',hue='sex',cumulative=False,stat='count',multiple='dodge')
plt.show()
```

```python
sns.countplot(data=tips,x='day',hue='sex')
plt.show()
```

```python
plt.figure(figsize=(6,6))
sns.stripplot(data=tips,x='day',y='tip',hue='sex',dodge=True)
plt.show()
```

```python
sns.catplot(x='day',hue='smoker',col='time',kind='count',data=tips)
plt.show()
```

```python
plt.figure(figsize=(6,6))
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='day',style='time',size='sex')
plt.show()
```

# JOINPLOT Y PAIRPLOT

```python
sns.jointplot(data=tips,x='total_bill',y='tip',hue='sex',kind='hist')
plt.show()
```

```python
sns.pairplot(data=tips,hue='sex',corner=True)
plt.show()
```
