```python
import keras
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
red = keras.models.load_model('/content/modelo_perrosygatos.keras')
red
```

```python
from keras.preprocessing import image
path_imagen_prueba = '/content/perro.jpeg'
imagen_prueba = image.load_img(path_imagen_prueba,target_size=(64,64))

test_image = image.img_to_array(imagen_prueba)
test_image = np.expand_dims(test_image,axis=0)


result = red.predict(test_image)
print("predicción : ",result)
if result[0][0] == 0:
  print("Es un gato")
  plt.imshow(imagen_prueba)
  plt.title('Gato')
  plt.axis('off')
  plt.show()
else:
  print("Es un perro")
  plt.imshow(imagen_prueba)
  plt.title('Perro')
  plt.axis('off')
  plt.show()
```
