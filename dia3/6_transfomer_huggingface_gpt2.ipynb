```python
from transformers import pipeline

text_generator = pipeline('text-generation', model='gpt2')
```

```python
output = text_generator('a caballo regalado', max_new_tokens=50)
print(output[0]['generated_text'])
```

```python
output = text_generator('in the future, ai will be', max_new_tokens=50)
print(output[0]['generated_text'])
```

```python
qa_pipeline = pipeline('question-answering')
contexto = "Tecsup es un instituto lider en Perú"
pregunta = "¿Qué es Tecsup?"
respuesta = qa_pipeline(question=pregunta, context=contexto)
print(respuesta)
```

```python
summarizer = pipeline('summarization', model='sshleifer/distilbart-cnn-12-6')
texto_largo = '''TECSUP es una institución educativa peruana especializada en la formación técnica y tecnológica,
reconocida por su fuerte enfoque práctico y su estrecha relación con la industria. Fundada en 1984,
prepara técnicos y profesionales en áreas clave como ingeniería, electricidad, electrónica, mecánica, automatización, minería y tecnologías de la información.
Su modelo educativo prioriza laboratorios, proyectos reales y el contacto directo con empresas,
lo que facilita una rápida inserción laboral. Cuenta con sedes en Lima, Arequipa y Trujillo, atendiendo a estudiantes de todo el país. En general,
TECSUP se distingue por formar profesionales altamente demandados y alineados con las necesidades reales del mercado laboral.'''
resumen = summarizer(texto_largo, max_length=50, min_length=25, do_sample=False)
print(resumen[0]['summary_text'])
```
