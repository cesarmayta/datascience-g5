# EMBEDDINGS

## ¿Qué son los Embeddings?

En **NLP (Natural Language Processing)**, las palabras no pueden procesarse directamente como texto. Por ello, se convierten en **vectores numéricos** llamados **embeddings**.

Los **embeddings** son representaciones densas y continuas en espacio vectorial, donde palabras similares están cerca unas de otras.

---

## 🎯 ¿Por qué usar Embeddings?

- Representan **relaciones semánticas** y **sintácticas** entre palabras.
- Reducen la **dimensionalidad** respecto al one-hot encoding.
- Permiten que los modelos de Deep Learning comprendan mejor el lenguaje natural.


## EJEMPLO SENCILLO PARA CLASIFICAR COMENTARIOS EN POSITIVOS Y NEGATIVOS

```python
import numpy as np
comentarios = [
    'me encanta este producto',
    'es muy malo y defectuoso',
    'funciona perfectamente bien',
    'la calidad es terrible',
    'excelente compra, lo recomiendo',
    'no sirve, una pérdida de dinero'
]
sentimientos = np.array([1, 0, 1, 0, 1, 0])
```

# RED NEURONAL PARA CLASIFICAR EL TEXTO CON ONE HOT ENCODING

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

# Tokenización One-Hot Encoding

```python
tokenizer = Tokenizer(num_words=1000)
tokenizer.fit_on_texts(comentarios)
sequences = tokenizer.texts_to_matrix(comentarios, mode='binary')
sequences.shape
```

# CREAMOS LA RED NEURONAL PARA PROCESAR LOS VECTORES SIN EMBEDDINGS

```python
modelo_sin_embeddings = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation='relu', input_shape=(sequences.shape[1],)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

modelo_sin_embeddings.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
modelo_sin_embeddings.summary()

# Entrenar
modelo_sin_embeddings.fit(sequences, sentimientos, epochs=20)
```

# CREAMOS UN RED NEURONAL PARA CLASIFICAR USANDO EMBEDDINGS

```python
# Re-tokenización para Embeddings (usa índices en lugar de vectores one-hot)
sequences = tokenizer.texts_to_sequences(comentarios)
padded_sequences = pad_sequences(sequences, maxlen=10)

# Modelo con embeddings
modelo_con_embeddings = tf.keras.Sequential([
    tf.keras.layers.Embedding(input_dim=1000, output_dim=16, input_length=10),
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

modelo_con_embeddings.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
modelo_con_embeddings.summary()

# Entrenar
modelo_con_embeddings.fit(padded_sequences, sentimientos, epochs=20)
```
