# CALCULO DE GRADIENTE

# CON ARRAY DE NUMPY

```python
import numpy as np
```

```python
x_np = np.array([1.0, 2.0, 3.0])
y_np = x_np ** 2

print("Resultado NumPy:", y_np)
```

## CON TENSORES DE PYTORCH

```python
x_torch = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
y_torch = x_torch ** 2

# Backpropagation
y_torch.sum().backward()

print("Resultado PyTorch:", y_torch)
print("Gradientes:", x_torch.grad)
```

# USO DE GPU

```python
# PyTorch puede usar GPU si está disponible
device = "cuda" if torch.cuda.is_available() else "cpu"
tensor_gpu = torch.tensor([1, 2, 3]).to(device)

print("PyTorch device:", tensor_gpu.device)
```

# CONVIRTIENDO UN ARRAY DE NUMPY A TENSOR DE PYTORCH

```python
np_array = np.array([1.0, 2.0, 3.0])
tensor_from_np = torch.from_numpy(np_array)
print(tensor_from_np)
```

# CONVIRTIENDO DE PYTORCH A NUMPY

```python
np_from_tensor = tensor_from_np.numpy()
print(np_from_tensor)
```

## 🔍 Diferencias entre NumPy y PyTorch

| Característica | NumPy (`ndarray`) | PyTorch (`Tensor`) |
|--------------|------------------|-------------------|
| Tipo de dato principal | `numpy.ndarray` | `torch.Tensor` |
| Operaciones matemáticas | ✅ Sí | ✅ Sí |
| Autograd (gradientes) | ❌ No | ✅ Sí |
| Backpropagation | ❌ No | ✅ Sí |
| Uso de GPU | ❌ No | ✅ Sí (`cuda`) |
| Deep Learning | ❌ No | ✅ Sí |
| Redes neuronales | ❌ No | ✅ Sí |
| Compatibilidad con CPU | ✅ Sí | ✅ Sí |
| Compatibilidad con GPU | ❌ No | ✅ Sí |
| Librerías asociadas | SciPy, Pandas | torch.nn, torch.optim |
| Conversión entre formatos | — | Compatible con NumPy |
| Uso principal | Cálculo numérico y análisis de datos | Machine Learning y Deep Learning |
| Framework de IA | ❌ No | ✅ Sí |
| Entrenamiento de modelos | ❌ No | ✅ Sí |

---

## 📌 Conclusión

- **NumPy** se usa principalmente para **cálculo numérico y análisis de datos**.
- **PyTorch** está diseñado para **entrenar modelos de Machine Learning y Deep Learning**.
- **PyTorch = NumPy + gradientes + GPU + redes neuronales** 🚀
